```python
import pandas as pd

df = pd.read_csv(r"D:\Code\Project\EMIPredict-AI\data\emi_cleaned.csv")
print(df.shape)
df.head()
```

```python
categorical_cols = ['gender', 'marital_status', 'education', 'employment_type', 'company_type', 
                     'house_type', 'existing_loans', 'emi_scenario']

df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
df_encoded.shape
```

```python
df['total_monthly_expense'] = (
    df['school_fees'] +
    df['college_fees'] +
    df['travel_expenses'] +
    df['groceries_utilities'] +
    df['other_monthly_expenses']
)
```

```python
df['debt_to_income'] = (df['current_emi_amount'] + df['monthly_rent']) / df['monthly_salary']
df['expense_to_income'] = df['total_monthly_expense'] / df['monthly_salary']
df['affordability_ratio'] = df['monthly_salary'] - df['total_monthly_expense'] - df['current_emi_amount'] - df['monthly_rent']

df[['debt_to_income', 'expense_to_income', 'affordability_ratio']].describe()
```

```python
categorical_cols = ['gender', 'marital_status', 'education', 'employment_type', 'company_type', 
                     'house_type', 'existing_loans', 'emi_scenario']

df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
df_encoded.shape
```

```python
from sklearn.preprocessing import StandardScaler

numeric_features = ['age', 'monthly_salary', 'years_of_employment', 'monthly_rent', 'family_size',
                     'dependents', 'total_monthly_expense', 'current_emi_amount', 'credit_score',
                     'bank_balance', 'emergency_fund', 'requested_amount', 'requested_tenure',
                     'debt_to_income', 'expense_to_income', 'affordability_ratio']

scaler = StandardScaler()
df_encoded[numeric_features] = scaler.fit_transform(df_encoded[numeric_features])
df_encoded[numeric_features].describe()
```

```python
from sklearn.model_selection import train_test_split

X = df_encoded.drop(columns=['emi_eligibility', 'max_monthly_emi'])
y_class = df_encoded['emi_eligibility']
y_reg = df_encoded['max_monthly_emi']

X_train, X_test, y_class_train, y_class_test, y_reg_train, y_reg_test = train_test_split(
    X, y_class, y_reg, test_size=0.2, random_state=42, stratify=y_class
)

print(X_train.shape, X_test.shape)
```

```python
mlflow.set_tracking_uri("sqlite:///D:/Code/Project/EMIPredict-AI/mlruns/mlflow.db")

experiment_name = "EMIPredict_Classification_final"
mlflow.create_experiment(
    experiment_name,
    artifact_location="file:///D:/Code/Project/EMIPredict-AI/mlruns"
)
mlflow.set_experiment(experiment_name)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

with mlflow.start_run(run_name="Logistic_Regression"):
    lr = LogisticRegression(max_iter=1000, random_state=42)
    lr.fit(X_train, y_class_train)
    
    y_pred = lr.predict(X_test)
    
    acc = accuracy_score(y_class_test, y_pred)
    prec = precision_score(y_class_test, y_pred, average='weighted')
    rec = recall_score(y_class_test, y_pred, average='weighted')
    f1 = f1_score(y_class_test, y_pred, average='weighted')
    
    mlflow.log_param("model", "LogisticRegression")
    mlflow.log_metric("accuracy", acc)
    mlflow.log_metric("precision", prec)
    mlflow.log_metric("recall", rec)
    mlflow.log_metric("f1_score", f1)
    mlflow.sklearn.log_model(lr, "model")
    
    print(f"Accuracy: {acc:.4f}, Precision: {prec:.4f}, Recall: {rec:.4f}, F1: {f1:.4f}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# --- Random Forest (baseline) ---
with mlflow.start_run(run_name="Random_Forest"):
    rf = RandomForestClassifier(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
    rf.fit(X_train, y_class_train)
    y_pred = rf.predict(X_test)

    mlflow.log_param("model", "RandomForest")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 15)
    mlflow.log_metric("accuracy", accuracy_score(y_class_test, y_pred))
    mlflow.log_metric("precision", precision_score(y_class_test, y_pred, average='weighted'))
    mlflow.log_metric("recall", recall_score(y_class_test, y_pred, average='weighted'))
    mlflow.log_metric("f1_score", f1_score(y_class_test, y_pred, average='weighted'))
    mlflow.sklearn.log_model(rf, "model")

    print("=== Random Forest (baseline) ===")
    print(classification_report(y_class_test, y_pred))

# --- Random Forest (class-balanced) ---
with mlflow.start_run(run_name="Random_Forest_Balanced"):
    rf_bal = RandomForestClassifier(n_estimators=100, max_depth=15, class_weight='balanced', random_state=42, n_jobs=-1)
    rf_bal.fit(X_train, y_class_train)
    y_pred_bal = rf_bal.predict(X_test)

    mlflow.log_param("model", "RandomForest_Balanced")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 15)
    mlflow.log_param("class_weight", "balanced")
    mlflow.log_metric("accuracy", accuracy_score(y_class_test, y_pred_bal))
    mlflow.log_metric("precision", precision_score(y_class_test, y_pred_bal, average='weighted'))
    mlflow.log_metric("recall", recall_score(y_class_test, y_pred_bal, average='weighted'))
    mlflow.log_metric("f1_score", f1_score(y_class_test, y_pred_bal, average='weighted'))
    mlflow.sklearn.log_model(rf_bal, "model")

    print("=== Random Forest (class-balanced) ===")
    print(classification_report(y_class_test, y_pred_bal))
```

```python
import xgboost as xgb
import mlflow.xgboost
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

le = LabelEncoder()
y_class_train_enc = le.fit_transform(y_class_train)
y_class_test_enc = le.transform(y_class_test)

sample_weights = compute_sample_weight(class_weight='balanced', y=y_class_train_enc)

with mlflow.start_run(run_name="XGBoost_Balanced"):
    xgb_clf = xgb.XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.1,
        random_state=42,
        n_jobs=-1,
        eval_metric='mlogloss'
    )
    xgb_clf.fit(X_train, y_class_train_enc, sample_weight=sample_weights)
    y_pred_xgb_enc = xgb_clf.predict(X_test)

    mlflow.log_param("model", "XGBoost_Balanced")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 6)
    mlflow.log_param("learning_rate", 0.1)
    mlflow.log_param("class_weight", "balanced (via sample_weight)")
    mlflow.log_metric("accuracy", accuracy_score(y_class_test_enc, y_pred_xgb_enc))
    mlflow.log_metric("precision", precision_score(y_class_test_enc, y_pred_xgb_enc, average='weighted'))
    mlflow.log_metric("recall", recall_score(y_class_test_enc, y_pred_xgb_enc, average='weighted'))
    mlflow.log_metric("f1_score", f1_score(y_class_test_enc, y_pred_xgb_enc, average='weighted'))
    mlflow.xgboost.log_model(xgb_clf, "model")

    print("=== XGBoost (class-balanced) ===")
    print(classification_report(y_class_test_enc, y_pred_xgb_enc, target_names=le.classes_))
```

```python
from sklearn.model_selection import train_test_split

y_reg = df_encoded['max_monthly_emi']

y_reg_train, y_reg_test = train_test_split(
    y_reg, test_size=0.2, stratify=y_class, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# ---------- 1. TRAIN LINEAR REGRESSION ----------
with mlflow.start_run(run_name="Linear_Regression"):
    lr_reg = LinearRegression()
    lr_reg.fit(X_train, y_reg_train)
    y_pred_lr = lr_reg.predict(X_test)

    mlflow.log_param("model", "LinearRegression")
    mlflow.log_metric("rmse", np.sqrt(mean_squared_error(y_reg_test, y_pred_lr)))
    mlflow.log_metric("mae", mean_absolute_error(y_reg_test, y_pred_lr))
    mlflow.log_metric("r2", r2_score(y_reg_test, y_pred_lr))
    mlflow.sklearn.log_model(lr_reg, "model")

# ---------- 2. TRAIN RANDOM FOREST REGRESSOR ----------
with mlflow.start_run(run_name="Random_Forest_Regressor"):
    rf_reg = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
    rf_reg.fit(X_train, y_reg_train)
    y_pred_rf = rf_reg.predict(X_test)

    mlflow.log_param("model", "RandomForestRegressor")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 15)
    mlflow.log_metric("rmse", np.sqrt(mean_squared_error(y_reg_test, y_pred_rf)))
    mlflow.log_metric("mae", mean_absolute_error(y_reg_test, y_pred_rf))
    mlflow.log_metric("r2", r2_score(y_reg_test, y_pred_rf))
    mlflow.sklearn.log_model(rf_reg, "model")

# ---------- 3. TRAIN XGBOOST REGRESSOR ----------
with mlflow.start_run(run_name="XGBoost_Regressor"):
    xgb_reg = xgb.XGBRegressor(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=42, n_jobs=-1)
    xgb_reg.fit(X_train, y_reg_train)
    y_pred_xgb = xgb_reg.predict(X_test)

    mlflow.log_param("model", "XGBoostRegressor")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 6)
    mlflow.log_param("learning_rate", 0.1)
    mlflow.log_metric("rmse", np.sqrt(mean_squared_error(y_reg_test, y_pred_xgb)))
    mlflow.log_metric("mae", mean_absolute_error(y_reg_test, y_pred_xgb))
    mlflow.log_metric("r2", r2_score(y_reg_test, y_pred_xgb))
    mlflow.xgboost.log_model(xgb_reg, "model")

# ---------- 4. EVALUATION FUNCTION ----------
def evaluate_regression(y_true, y_pred, model_name):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)

    print(f"\n{model_name}")
    print("RMSE :", rmse)
    print("MAE  :", mae)
    print("R²   :", r2)

# ---------- 5. MODEL EVALUATION ----------
y_test_r = y_reg_test

evaluate_regression(y_test_r, y_pred_lr, "Linear Regression")
evaluate_regression(y_test_r, y_pred_rf, "Random Forest Regressor")
evaluate_regression(y_test_r, y_pred_xgb, "XGBoost Regressor")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
)

# Encode labels for models that need numeric classes (XGBoost)
le = LabelEncoder()
y_train_enc = le.fit_transform(y_class_train)
y_test_enc = le.transform(y_class_test)

def evaluate_model(y_true, y_pred, y_prob, model_name):
    print(f"\n{model_name}")
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred, average='weighted'))
    print("Recall   :", recall_score(y_true, y_pred, average='weighted'))
    print("F1 Score :", f1_score(y_true, y_pred, average='weighted'))
    print("ROC-AUC  :", roc_auc_score(y_true, y_prob, multi_class='ovr'))

# ---------- LOGISTIC REGRESSION ----------
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_class_train)
y_pred_lr = log_reg.predict(X_test)
y_prob_lr = log_reg.predict_proba(X_test)
evaluate_model(y_class_test, y_pred_lr, y_prob_lr, "Logistic Regression")

# ---------- RANDOM FOREST ----------
rf_clf = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42, n_jobs=-1)
rf_clf.fit(X_train, y_class_train)
y_pred_rf = rf_clf.predict(X_test)
y_prob_rf = rf_clf.predict_proba(X_test)
evaluate_model(y_class_test, y_pred_rf, y_prob_rf, "Random Forest")

# ---------- XGBOOST ----------
xgb_clf = XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=6,
    subsample=0.8, colsample_bytree=0.8,
    objective='multi:softprob', eval_metric='mlogloss',
    random_state=42, n_jobs=-1
)
xgb_clf.fit(X_train, y_train_enc)
y_pred_xgb_enc = xgb_clf.predict(X_test)
y_prob_xgb = xgb_clf.predict_proba(X_test)
y_pred_xgb = le.inverse_transform(y_pred_xgb_enc)
evaluate_model(y_class_test, y_pred_xgb, y_prob_xgb, "XGBoost")
```
