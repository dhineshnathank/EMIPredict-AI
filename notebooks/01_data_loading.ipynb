```python
import pandas as pd

df = pd.read_csv(r"D:\Code\Project\EMIPredict-AI\data\emi_prediction_dataset.csv")
print(df.shape)
df.head()
```

```python
print(df['age'].unique()[:20])
print(df['monthly_salary'].unique()[:20])
print(df['bank_balance'].unique()[:20])
```

```python
for col in ['age', 'monthly_salary', 'bank_balance']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[['age', 'monthly_salary', 'bank_balance']].info()
```

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
(df.isnull().sum() / len(df) * 100).round(2)[lambda x: x > 0]
```

```python
num_cols = ['age', 'monthly_salary', 'monthly_rent', 'credit_score', 'bank_balance', 'emergency_fund']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

df['education'] = df['education'].fillna(df['education'].mode()[0])

df.isnull().sum().sum()
```
